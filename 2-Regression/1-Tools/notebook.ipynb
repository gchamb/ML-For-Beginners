## Diabetes Dataset Practice

```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn import datasets, linear_model, model_selection
```

```python
# Load the diabetes dataset
x, y = datasets.load_diabetes(return_X_y=True)

print("Features shape:", x.shape)

blood_pressure = x[:, 3]  # Use only the fourth feature (bp)
blood_pressure = blood_pressure.reshape(-1, 1)  # Reshape for sklearn

print("Blood Pressure shape:", blood_pressure.shape, blood_pressure[0])
```

```python
# Split data into training and testing sets

x_train, x_test, y_train, y_test = model_selection.train_test_split(
    blood_pressure, y, test_size=0.33,
)

# Create and train the model
model = linear_model.LinearRegression()
model.fit(x_train, y_train)

# Make predictions
y_pred = model.predict(x_test)
```

```python
plt.scatter(x_test, y_test,  color='black')
plt.plot(x_test, y_pred, color='blue', linewidth=3)
plt.xlabel('Scaled Blood Pressure')
plt.ylabel('Disease Progression')
plt.title('A Graph Plot Showing Diabetes Progression Against BMI')
plt.show()
```
